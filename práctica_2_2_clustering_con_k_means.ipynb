# **K-means**

**K-means (k-medias)** es un algoritmo de aprendizaje automático no supervisado que divide un conjunto de datos en k grupos o clústeres basándose en la similitud de sus características.

Su principal limitación es que el usuario debe definir el valor de k inicialmente.

Este algoritmo se usa cuando tenemos datos sin etiquetas. Por ejemplo, medidas de flores (largo de pétalo, ancho de pétalo) pero no se sabe qué tipos de flores son. Se quiere descubrir cuántos tipos de flores hay.

En realidad, sí se conocen las tres especies, que luego usaremos para comparar el resultado obtenido: Iris **setosa**, Iris **versicolor** e Iris **virginica**.

```python
from sklearn.datasets import load_iris

iris = load_iris()
print(type(iris))
```

```python
iris.feature_names
iris.target
#iris.target_names
#iris.DESCR
```

```python
datos = iris.data
print(type(datos))
datos
```

```python
# operaciones
primera_flor = datos[0]
primera_flor
un_valor = datos[5, 2]
un_valor
rango10a15 = datos[10:15]
rango10a15
rango0a3 = datos[:3]
rango0a3
```

```python
# se dispone de 150 datos, para cada uno 4 medidas
datos.shape
```

```python
from sklearn.cluster import KMeans

# === este es el número de grupos ===
n_clusters = 3
# =====================================

# se configura el modelo
modelo = KMeans(n_clusters=n_clusters, random_state=0, n_init=10)
# se entrena el modelo
modelo.fit(datos)
```

```python
print(modelo.labels_)          # etiqueta de cluster (0, 1, 2...) de cada muestra
print(modelo.cluster_centers_) # coordenadas de los K centroides (shape: K × n_features)
print(modelo.inertia_)         # suma de distancias² (cómo de "compactos" son los clusters)
print(modelo.n_iter_)          # iteraciones que necesitó para converger
```

```python
import matplotlib.pyplot as plt

plt.scatter(datos[:, 0], datos[:, 1], c=modelo.labels_, cmap="viridis")
plt.title(f"Agrupación con {n_clusters} grupos")
plt.show()
```

```python
# prueba el código con 2 grupos
n_clusters = 2

modelo = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
modelo.fit(datos)

plt.scatter(datos[:, 0], datos[:, 1], c=modelo.labels_, cmap="viridis")
plt.title(f"Agrupación con {n_clusters} grupos")
plt.show()
```

```python
# prueba el código con 4 grupos
n_clusters = 4

modelo = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
modelo.fit(datos)

plt.scatter(datos[:, 0], datos[:, 1], c=modelo.labels_, cmap="viridis")
plt.title(f"Agrupación con {n_clusters} grupos")
plt.show()
```

```python
# prueba el código con 5 grupos
n_clusters = 5

modelo = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
modelo.fit(datos)

plt.scatter(datos[:, 0], datos[:, 1], c=modelo.labels_, cmap="viridis")
plt.title(f"Agrupación con {n_clusters} grupos")
plt.show()
```

```python
# aquí va la implementación
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import adjusted_rand_score, confusion_matrix

# Número de grupos
n_clusters = 3

# Guardamos las etiquetas reales y las predichas
etiquetas_reales = iris.target
etiquetas_kmeans = modelo.labels_

# Creamos una figura con dos gráficos
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

# Gráfico de las especies reales
scatter1 = ax1.scatter(
    tabla['sepal length (cm)'],
    tabla['sepal width (cm)'],
    c=etiquetas_reales,
    cmap='viridis'
)

ax1.set_title("Especies reales")
ax1.set_xlabel("Longitud del sépalo (cm)")
ax1.set_ylabel("Anchura del sépalo (cm)")
ax1.legend(
    handles=scatter1.legend_elements()[0],
    labels=list(iris.target_names),
    title="Especies"
)

# Gráfico de los grupos encontrados por K-Means
scatter2 = ax2.scatter(
    tabla['sepal length (cm)'],
    tabla['sepal width (cm)'],
    c=etiquetas_kmeans,
    cmap='viridis'
)

ax2.set_title(f"K-Means ({n_clusters} grupos)")
ax2.set_xlabel("Longitud del sépalo (cm)")
ax2.set_ylabel("Anchura del sépalo (cm)")
ax2.legend(
    handles=scatter2.legend_elements()[0],
    labels=[f'Grupo {i}' for i in range(n_clusters)],
    title="Clústeres"
)

plt.suptitle(
    "Comparación: especies reales vs. K-Means",
    fontsize=14,
    fontweight='bold'
)
plt.tight_layout()
plt.show()

# Mostramos las etiquetas obtenidas
print("Etiquetas reales:")
print(etiquetas_reales)

print("\nEtiquetas obtenidas por K-Means:")
print(etiquetas_kmeans)

# Comparamos los resultados mediante una matriz de contingencia
comparacion = pd.crosstab(
    pd.Series(etiquetas_reales, name="Especie real"),
    pd.Series(etiquetas_kmeans, name="Grupo K-Means")
)
```
